```python
daily_sales = [
    2300, 1950, 2800, 2100, 3600, 5200, 4100,
    2500, 2200, 3100, 2400, 3900, 5600, 4700
]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg)
# cost and price are in KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140}
}
```

## 1. Overall Sales Performance

This section calculates total, average, highest and lowest daily sales
over the 14-day period. It also identifies the day and week in which
the highest and lowest sales occurred.

```python
# Calculate overall sales statistics

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the positions of the highest and lowest sales
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

# Identify the corresponding days
best_day = days[best_index]
worst_day = days[worst_index]

# Determine the week numbers
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

print("OVERALL SALES PERFORMANCE")
print("-" * 35)
print(f"Total Sales:       KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales:     KES {highest_sales:,.2f}")
print(f"Lowest Sales:      KES {lowest_sales:,.2f}")
print(f"Best Sales Day:    {best_day}, Week {best_week}")
print(f"Worst Sales Day:   {worst_day}, Week {worst_week}")
```

## 2. Weekly Sales Comparison

This section compares sales in Week 1 and Week 2 and calculates
the percentage growth or decline between the two weeks.

```python
def week_totals(sales):
    """
    Calculate the total sales for the first and second weeks.

    Args:
        sales: A list containing 14 daily sales values.

    Returns:
        A tuple containing Week 1 and Week 2 sales totals.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return week1, week2


# Unpack the returned tuple
week1_total, week2_total = week_totals(daily_sales)

# Calculate week-on-week growth percentage
growth_percentage = (
    (week2_total - week1_total) / week1_total
) * 100

print("WEEKLY SALES COMPARISON")
print("-" * 35)
print(f"Week 1 Total: KES {week1_total:,.2f}")
print(f"Week 2 Total: KES {week2_total:,.2f}")
print(f"Week-on-Week Growth: {growth_percentage:.2f}%")
```

## 3. Daily Sales Classification

Each day is classified according to how its sales compare with
the 14-day average.

- Strong: More than 20% above the average.
- Weak: More than 20% below the average.
- Normal: Within the remaining range.

```python
def classify_day(amount, average):
    """
    Classify daily sales based on the overall average.

    Args:
        amount: Sales amount for a particular day.
        average: Average sales for the 14-day period.

    Returns:
        Strong, Weak or Normal.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Initialise the category counters
category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Classify every day's sales and count each category
daily_classifications = []

for day, amount in zip(days, daily_sales):
    category = classify_day(amount, average_sales)
    daily_classifications.append(category)
    category_counts[category] += 1

    print(f"{day}: KES {amount:,.2f} - {category}")


print("\nSALES CATEGORY SUMMARY")
print("-" * 35)

for category, count in category_counts.items():
    print(f"{category}: {count} days")
```

## 4. Weekday Versus Weekend Sales

This section compares average daily sales from Monday to Friday
with average daily sales on Saturday and Sunday.

```python
# Separate weekday and weekend sales
weekday_sales = []
weekend_sales = []

for index, amount in enumerate(daily_sales):
    if index % 7 < 5:
        weekday_sales.append(amount)
    else:
        weekend_sales.append(amount)


# Calculate average sales for each group
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate the percentage difference relative to weekdays
weekend_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print("WEEKDAY VS WEEKEND PERFORMANCE")
print("-" * 40)
print(f"Average Weekday Sales: KES {weekday_average:,.2f}")
print(f"Average Weekend Sales: KES {weekend_average:,.2f}")
print(f"Weekend Sales Difference: {weekend_difference:.2f}%")
```

## 5. Stock Analysis and Gross Margins

This section identifies products whose available stock is at or
below their reorder levels. It also calculates the gross margin
percentage for each product.

```python
def products_to_reorder(stock_data):
    """
    Return products whose current stock is at or below reorder level.

    Args:
        stock_data: Dictionary containing product stock information.

    Returns:
        A list of product names requiring restocking.
    """
    reorder_list = []

    for product, details in stock_data.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


def calculate_gross_margin(cost, price):
    """
    Calculate gross margin as a percentage of selling price.

    Args:
        cost: Cost price per kg.
        price: Selling price per kg.

    Returns:
        Gross margin percentage.
    """
    return ((price - cost) / price) * 100


# Identify products requiring restocking
reorder_products = products_to_reorder(stock)

print("PRODUCTS REQUIRING RESTOCKING")
print("-" * 40)

for product in reorder_products:
    print(
        f"{product}: {stock[product]['qty']} kg "
        f"(Reorder level: {stock[product]['reorder']} kg)"
    )


# Calculate and display gross margins
print("\nPRODUCT GROSS MARGINS")
print("-" * 55)
print(f"{'Product':<15}{'Cost (KES)':>12}{'Price (KES)':>14}{'Margin':>12}")
print("-" * 55)

for product, details in stock.items():
    margin = calculate_gross_margin(
        details["cost"],
        details["price"]
    )

    print(
        f"{product:<15}"
        f"{details['cost']:>12,.2f}"
        f"{details['price']:>14,.2f}"
        f"{margin:>11.2f}%"
    )
```

```python
print("=" * 65)
print("                 STALL PERFORMANCE REPORT")
print("                    GIKOMBA MARKET")
print("=" * 65)

print("\n1. OVERALL SALES PERFORMANCE")
print("-" * 65)
print(f"{'Total Sales:':<30} KES {total_sales:>15,.2f}")
print(f"{'Average Daily Sales:':<30} KES {average_sales:>15,.2f}")
print(f"{'Highest Daily Sales:':<30} KES {highest_sales:>15,.2f}")
print(f"{'Lowest Daily Sales:':<30} KES {lowest_sales:>15,.2f}")
print(f"{'Best Sales Day:':<30} {best_day}, Week {best_week}")
print(f"{'Worst Sales Day:':<30} {worst_day}, Week {worst_week}")

print("\n2. WEEKLY SALES COMPARISON")
print("-" * 65)
print(f"{'Week 1 Sales:':<30} KES {week1_total:>15,.2f}")
print(f"{'Week 2 Sales:':<30} KES {week2_total:>15,.2f}")
print(f"{'Week-on-Week Growth:':<30} {growth_percentage:>14.2f}%")

print("\n3. DAILY SALES CLASSIFICATION")
print("-" * 65)

for category, count in category_counts.items():
    print(f"{category + ' Days:':<30} {count:>15}")

print("\n4. WEEKDAY AND WEEKEND PERFORMANCE")
print("-" * 65)
print(f"{'Average Weekday Sales:':<30} KES {weekday_average:>15,.2f}")
print(f"{'Average Weekend Sales:':<30} KES {weekend_average:>15,.2f}")
print(f"{'Weekend Difference:':<30} {weekend_difference:>14.2f}%")

print("\n5. STOCK REORDER REQUIREMENTS")
print("-" * 65)

for product in reorder_products:
    print(
        f"{product:<30} "
        f"{stock[product]['qty']} kg available | "
        f"Reorder level: {stock[product]['reorder']} kg"
    )

print("\n6. PRODUCT GROSS MARGINS")
print("-" * 65)
print(f"{'Product':<20}{'Gross Margin':>20}")
print("-" * 65)

for product, details in stock.items():
    margin = calculate_gross_margin(
        details["cost"],
        details["price"]
    )
    print(f"{product:<20}{margin:>19.2f}%")

print("\n" + "=" * 65)
print("                  END OF REPORT")
print("=" * 65)
```
